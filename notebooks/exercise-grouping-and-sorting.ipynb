```python
# importing pandas and the csv file
import pandas as pd

reviews = pd.read_csv("../input/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
#pd.set_option("display.max_rows", 5)

from learntools.core import binder; binder.bind(globals())
from learntools.pandas.grouping_and_sorting import *
print("Setup complete.")
```

```python
# The most common wine reviewers in the dataset
reviews_written = reviews.groupby("taster_twitter_handle").size()
# Check your answer

q1.check()
```

```python
# The best wine I can buy for a given amount of money
best_rating_per_price = reviews.groupby("price").points.max()

# Check your answer
q2.check()
```

```python
# The minimum and maximum prices for each variety of wine
price_extremes = reviews.groupby("variety").price.agg(["min", "max"])

# Check your answer
q3.check()
```

```python
# The most expensive wine varieties.

sorted_varieties = price_extremes.sort_values(by = ["min", "max"], ascending = False)

# Check your answer
q4.check()
```
